# 3. 本地模型
## 3.1 本地模型框架
LangChain也支持使用 Ollama 、 vLLM 等框架启动的本地大模型。

Ollama是在Github上的一个开源项目，其项目定位是：一个本地运行大模型的集成框架，可以实现如Qwen、Deepseek 等主流大模型的下载、启动和本地运行的自动化部署及推理流程。

## 3.2 Ollama 模型
Ollama项目支持跨平台部署，目前已兼容Mac、Linux和Windows操作系统。

无论使用哪个操作系统，Ollama项目的安装过程都设计得非常简单。

你可以通过命令行 和 GUI 进行安装Ollama所需要的模型。

## 3.3 LangChain调用模型
LangChain项目支持使用 Ollama 模型，也可以使用 vLLM 模型。不过LangChain可以整合 Ollama 模型，实现对本地模型的调用。

```python
#pip install langchain-ollama
pip install -qU langchain-ollama
pip install -U ollama
```

```python
from langchain_ollama import ChatOllama
ollama_llm = ChatOllama(
    model_name="deepseek-r1:1.5b",
    base_url="http://localhost:11434" #如果Ollama在本地默认端口运行，则可省略，或使用http://localhost:11434
)
question = "你好，请你介绍一下你自己。"
result = ollama_llm.invoke(question)
print(result)
```

不过，我们也可以通过统一接口的方式调用Ollama模型。

```python
from langchain.chat_models import init_chat_model
ollama_llm = init_chat_model(
    model="deepseek-r1:1.5b",
    model_provider="ollama",
    base_url="http://localhost:11434",
)
question = "你好，请你介绍一下你自己。"
result = ollama_llm.invoke(question)
print(result)
```
