## 7.1 画像字段说明

画像是一个字典（类型为 `ModelProfile`），描述模型"能做什么、上限多大"。下表按类别说明每个字段，并对比上面两个模型的值（"—"表示画像中没有这个字段）：

| 字段 | 含义 | GPT-6 Luna | DeepSeek V4 Flash |
|---|---|:---:|:---:|
| **模型信息** | | | |
| `name` | 模型的可读名称 | GPT-6 Luna | DeepSeek V4 Flash |
| `status` | 生命周期状态，如 `active`、`deprecated` | — | — |
| `release_date` | 发布日期 | 2026-09-22 | 2026-09-10 |
| `last_updated` | 数据最近更新的日期 | 2026-09-22 | 2026-09-10 |
| `open_weights` | 是否开放模型权重（开源） | False | True |
| **输入能力** | | | |
| `max_input_tokens` | 最大输入 token 数，即上下文窗口 | 1,050,000 | 1,000,000 |
| `text_inputs` | 支持文本输入 | True | True |
| `image_inputs` | 支持图片输入 | True | True |
| `image_url_inputs` | 支持通过 URL 传入图片 | True | — |
| `pdf_inputs` | 支持 PDF 输入 | True | — |
| `audio_inputs` | 支持音频输入 | False | False |
| `video_inputs` | 支持视频输入 | False | False |
| `image_tool_message` | 工具的返回结果（ToolMessage）中可以包含图片 | True | — |
| `pdf_tool_message` | 工具的返回结果中可以包含 PDF | True | — |
| **输出能力** | | | |
| `max_output_tokens` | 最大输出 token 数 | 128,000 | 384,000 |
| `text_outputs` | 能输出文本 | True | True |
| `image_outputs` / `audio_outputs` / `video_outputs` | 能输出图片 / 音频 / 视频 | False | False |
| `reasoning_output` | 支持推理（会输出思考过程） | True | True |
| `reasoning_effort_levels` | 可选的思考力度，如 `['low', 'medium', 'high']` | — | — |
| `reasoning_effort_default` | 默认的思考力度 | — | — |
| **工具调用** | | | |
| `tool_calling` | 支持工具调用 | True | True |
| `tool_choice` | 支持强制模型调用指定的工具 | True | — |
| `tool_call_streaming` | 流式输出时能正确返回工具调用的片段 | True | True |
| **其他** | | | |
| `structured_output` | 支持厂商原生的结构化输出 | True | True |
| `attachment` | 支持文件附件 | True | True |
| `temperature` | 支持 temperature 参数 | False | True |

几点观察：

- **字段缺失 ≠ 不支持**。画像中的字段都是可选的，缺失只表示"没有这项数据"。读取时用 `profile.get("pdf_inputs")`，不要用 `profile["pdf_inputs"]`，否则可能报 `KeyError`。
- **GPT-6 Luna 的 `temperature` 为 False**：推理模型不支持调温度。画像只是一份"说明书"，LangChain 不会据此自动去掉你设置的 temperature，需要自己判断（做法见 7.3）。
- **DeepSeek 的 `temperature` 为 True，但思考模式下 temperature 并不生效**（见 `02-model-init-parameters.ipynb` 1.5 节）。画像描述的是"接口支持哪些参数"，不会细分到"某个模式下是否生效"。
- DeepSeek V4 Flash 的 `max_output_tokens` 为 384,000，与 DeepSeek 文档中 max_tokens 的上限（384K）一致。

## 7.2 画像数据从哪来？

画像**不是**调用模型时从厂商那里实时获取的，而是**内置在集成包里的静态数据**：

1. 数据来自开源项目 [models.dev](https://github.com/sst/models.dev)（收集各家模型的能力数据），LangChain 又补充了一些字段；
2. 这些数据被生成为集成包中的一个 Python 文件，例如 `langchain_deepseek/data/_profiles.py`；
3. 创建模型对象时，LangChain 用**模型名**去这份数据里**精确查找**：找到了就填入 `profile`，找不到就是 `None`（查找出错也不会影响模型的创建）。

因此：

- 画像是否存在，取决于**用的是哪个集成类**，以及**模型名是否在它的数据中**；
- 想获得新模型的画像，需要**升级集成包**；
- 画像是 **Beta 功能**，字段和格式将来可能变化。

下面验证几种情况（只创建模型对象，不调用 API）：

```python
from langchain.chat_models import init_chat_model
from langchain_openai import ChatOpenAI
from langchain_deepseek.data._profiles import _PROFILES  # 内置的画像数据（内部变量，仅用于查看）
from dotenv import load_dotenv
import os
load_dotenv(override=True, verbose=True,dotenv_path="/Users/skk/Developer/lang-chain-learn/conf/.env")

print("langchain-deepseek 内置了这些模型的画像：", list(_PROFILES))

# 1. ChatDeepSeek + 数据中有的模型名 → 有画像
model1 = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
)
# 2. 同一个模型，改用 ChatOpenAI 兼容方式创建 → 没有画像（langchain-openai 的数据里没有 DeepSeek 模型）
model2 = ChatOpenAI(
    model="deepseek-v4-flash",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
)
# 3. ChatDeepSeek + 数据中没有的模型名（DeepSeek 的旧模型名）→ 没有画像
model3 = init_chat_model(
    model="deepseek-chat",
    model_provider="deepseek",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
)
print("ChatDeepSeek + deepseek-v4-flash：", model1.profile["name"])
print("ChatOpenAI 兼容方式 + deepseek-v4-flash：", model2.profile)
print("ChatDeepSeek + deepseek-chat：", model3.profile)
```

## 7.3 画像有什么用？

**1. 在程序中按能力做判断**：比如先检查模型是否支持图片输入、输入是否超过上下文窗口，再决定怎么调用，避免运行时才报错。

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
load_dotenv(override=True, verbose=True,dotenv_path="/Users/skk/Developer/lang-chain-learn/conf/.env")

def check_model(model, need_image=False, input_tokens=0):
    """根据画像检查模型能否完成任务"""
    profile = model.profile or {}  # 画像可能为 None，先兜底成空字典
    if not profile:
        return "没有画像数据，无法判断"
    if need_image and not profile.get("image_inputs"):
        return "不支持图片输入"
    max_input = profile.get("max_input_tokens")
    if max_input and input_tokens > max_input:
        return f"输入 {input_tokens:,} 个 token，超过了上下文窗口 {max_input:,}"
    return "可以使用"

for name in ["deepseek-v4-flash", "deepseek-v4-pro"]:
    model = init_chat_model(
        model=name,
        model_provider="deepseek",
        api_key=os.getenv("DEEPSEEK_API_KEY"),
        base_url=os.getenv("DEEPSEEK_BASE_URL"),
    )
    print(f"{name}：识别图片 → {check_model(model, need_image=True)}；"
          f"读 120 万 token 的文档 → {check_model(model, input_tokens=1_200_000)}")
```

**2. LangChain 内部也会读取画像**（后面学 Agent 时会遇到）：

| 功能 | 读取的字段 | 作用 |
|---|---|---|
| `create_agent` 的结构化输出 | `structured_output` | 模型支持原生结构化输出时，自动使用厂商自带的方式，否则改用工具调用的方式 |
| 摘要中间件 `SummarizationMiddleware` | `max_input_tokens` | 按上下文窗口的比例（如用到 80%）触发对话历史压缩。画像中没有这个字段时，按比例的配置会直接报错 |

这正是画像存在的意义：让框架**根据模型的能力自动选择合适的做法**。

## 7.4 自定义或修正画像

以下情况需要自己提供画像：通过 ChatOpenAI 兼容方式或中转服务调用（画像为 None）、使用的新模型还没被收录、内置数据有误。有两种方式：

```python
from langchain.chat_models import init_chat_model
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os
load_dotenv(override=True, verbose=True,dotenv_path="/Users/skk/Developer/lang-chain-learn/conf/.env")

# 方式一：初始化时传入 profile —— 会“整体替换”默认画像（不是合并）
model1 = ChatOpenAI(
    model="deepseek-v4-flash",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
    profile={"max_input_tokens": 1_000_000, "tool_calling": True, "image_inputs": True},
)
print("方式一：", model1.profile)

# 方式二：在默认画像的基础上只修改部分字段（用 | 合并字典，再用 model_copy 生成新对象）
model2 = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
)
model3 = model2.model_copy(update={"profile": model2.profile | {"max_input_tokens": 128_000}})
print("方式二：原模型", model2.profile["max_input_tokens"], "→ 新模型", model3.profile["max_input_tokens"],
      "，其他字段仍然保留：", model3.profile["name"])
```

注意：

- 方式一会整体替换，没有传入的字段就没有了；只想改几个字段时用方式二。
- 字段名要写对。写了 `ModelProfile` 中没有定义的键时，LangChain 会给出警告 `Unrecognized keys in model profile`。

## 8.2 这些参数分别来自哪个类？

开头提到参数"由自身定义或从父类继承"。ChatDeepSeek 的继承关系是：

`ChatDeepSeek` → `BaseChatOpenAI` → `BaseChatModel` → `BaseLanguageModel` → `RunnableSerializable`

因为 ChatDeepSeek 是基于 OpenAI 兼容接口实现的，所以**绝大多数参数来自 `BaseChatOpenAI`**，而不是 BaseChatModel。下面沿着继承链，按"定义参数的类"分组统计：

```python
from langchain_deepseek import ChatDeepSeek

def defined_in(field_name):
    """沿着继承链（MRO）查找最先定义这个参数的类"""
    for cls in ChatDeepSeek.__mro__:
        if field_name in cls.__dict__.get("__annotations__", {}):
            return cls.__name__
    return "未知"

groups = {}
for field_name in ChatDeepSeek.model_fields:
    groups.setdefault(defined_in(field_name), []).append(field_name)

for cls_name, names in groups.items():
    print(f"{cls_name}（{len(names)} 个）：{', '.join(names)}\n")
```

- **RunnableSerializable、BaseLanguageModel、BaseChatModel**：LangChain 框架通用参数（见 8.4），所有聊天模型都有；
- **BaseChatOpenAI**：OpenAI 兼容接口的全部参数，ChatDeepSeek 直接继承过来，所以其中也包括一些 DeepSeek 用不到的 OpenAI 专用参数；
- **ChatDeepSeek**：只定义了 3 个。子类可以重新定义父类的参数，比如 `model_name` 在 BaseChatOpenAI 中有默认值 `gpt-3.5-turbo`，ChatDeepSeek 重新定义后变成了必填。上面的统计按最后定义它的类计算。

## 8.3 参数名与别名

`model_fields` 中列出的是**字段名**，而我们创建模型时写的往往是**别名**。比如传入的是 `model=`，存到的字段却是 `model_name`。可以用下面的代码查看所有带别名的参数：

```python
from langchain_deepseek import ChatDeepSeek

for field_name, field in ChatDeepSeek.model_fields.items():
    if field.alias:
        print(f"创建时写 {field.alias:15} → 存到字段 {field_name}")
```

几点说明：

- `base_url` 出现了两次：ChatDeepSeek 自己定义了 `api_base`，又从 BaseChatOpenAI 继承了 `openai_api_base`，传入的 `base_url` 会同时赋给这两个字段。`api_key` 也一样，会同时赋给 `api_key` 和 `openai_api_key`。
- `stop` 可以写成 `stop=` 或 `stop_sequences=`，`timeout` 对应的字段是 `request_timeout`。
- 不传 `api_key`、`base_url` 时，ChatDeepSeek 会读取环境变量：`api_key` 读 `DEEPSEEK_API_KEY`，`base_url` 读 **`DEEPSEEK_API_BASE`**（注意不是本项目 `.env` 中的 `DEEPSEEK_BASE_URL`）。两者都没有时，使用官方地址 `https://api.deepseek.com/v1`。

## 8.4 全部参数分类速查

除了 `01-model-init-online.ipynb`、`02-model-init-parameters.ipynb` 中讲过的**连接参数**和**模型推理参数**，还有一类 **LangChain 框架通用参数**，所有聊天模型都有：

| 参数 | 作用 | 默认值 |
|---|---|---|
| `name` | 模型对象的显示名称，用于在追踪记录（如 LangSmith）中区分不同对象。**不是模型名称** | `None`（显示为类名） |
| `verbose` | 是否打印详细日志。单独调用模型时基本看不到效果，主要在链、Agent 中打印执行过程 | `False` |
| `callbacks` | 回调处理器列表，用于集成 LangSmith 或自定义监控（如 04 笔记中的计时回调） | `None` |
| `tags` / `metadata` | 该模型对象的标签和元数据，会写入每次调用的运行记录（与 config 中同名键的区别见 9.1） | `None` |
| `cache` | 是否缓存调用结果：`True` 使用全局缓存，`False` 不缓存，`None` 有全局缓存就用；也可以直接传入缓存对象。流式调用不支持缓存 | `None` |
| `rate_limiter` | LangChain 内置的限流器，控制请求频率（用法见 02 笔记第 6 节） | `None` |
| `disable_streaming` | 禁用流式输出：`True` 时 `stream()` 退化为 `invoke()`；`"tool_calling"` 表示只在绑定了工具时禁用 | `False` |
| `output_version` | AIMessage 的内容格式：`"v0"` 为厂商原始格式，`"v1"` 为 LangChain 统一的内容块格式 | `None`（相当于 v0） |
| `custom_get_token_ids` | 自定义分词函数，用于在本地统计 token 数 | `None` |
| `profile` | 模型能力画像（第 7 节） | 自动从集成包加载 |

以 `cache` 为例：开启缓存后，**完全相同**的请求第二次会直接返回缓存结果，不再调用 API：

```python
import time
from langchain_core.caches import InMemoryCache
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
load_dotenv(override=True, verbose=True,dotenv_path="/Users/skk/Developer/lang-chain-learn/conf/.env")

model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
    cache=InMemoryCache(),  # 使用内存缓存（程序结束后清空）
)

for i in range(2):
    start = time.perf_counter()
    response = model.invoke("用一句话介绍 LangChain")
    print(f"第 {i + 1} 次：耗时 {time.perf_counter() - start:.3f} 秒，usage_metadata = {response.usage_metadata}")
```

第二次几乎不花时间，因为结果直接来自缓存。第二次的 `usage_metadata` 多了 `'total_cost': 0`，这是 LangChain 给缓存结果加的标记（源码注释：命中缓存时把费用清零）。其中的 token 数只是第一次调用的记录，并没有再次消耗。

缓存的键是"提示词 + 模型参数"，提示词哪怕多一个空格也不会命中。适合开发调试时反复运行同样的代码，或者重复问题很多的场景（如 FAQ）。

其余参数按用途分类如下：

**连接参数**（`01`、`02` 笔记中已讲过大部分）

| 参数 | 作用 |
|---|---|
| `model_name`（创建时写 `model`） | 模型名称 |
| `api_key` / `openai_api_key` | API 密钥 |
| `api_base` / `openai_api_base`（创建时写 `base_url`） | API 地址 |
| `openai_organization`、`openai_proxy` | OpenAI 的组织 ID、代理地址 |
| `request_timeout`（创建时写 `timeout`）、`max_retries` | 超时时间、最大重试次数 |
| `default_headers`、`default_query` | 每次请求都附带的 HTTP 请求头、URL 查询参数 |

**生成（推理）参数**（`02` 笔记中已讲过大部分）

| 参数 | 作用 |
|---|---|
| `temperature`、`top_p` | 采样的随机性 |
| `max_tokens`、`stop` | 最大输出长度、停止序列 |
| `presence_penalty`、`frequency_penalty` | 重复惩罚（DeepSeek V4 已废弃） |

| `n` | 一次生成几个候选回答（需要厂商支持） |
| `logprobs`、`top_logprobs` | 返回 token 的概率 |
| `logit_bias` | 调整指定 token 出现的概率 |
| `reasoning_effort` | 思考力度 |
| `streaming`、`stream_usage` | 是否默认流式输出、流式输出时是否返回 token 用量 |

**扩展参数**

| 参数 | 作用 |
|---|---|
| `extra_body` | 厂商独有的参数（见下方说明） |
| `model_kwargs` | OpenAI 规范中有、但 LangChain 没有单独定义的参数（见下方说明） |
| `disabled_params` | 禁止发送某些参数，用于不支持这些参数的模型 |
| `include_response_headers` | 是否把 HTTP 响应头放进 `response_metadata` |

**底层客户端**（一般不用设置，LangChain 会根据上面的参数自动创建）

| 参数 | 作用 |
|---|---|
| `client`、`async_client`、`root_client`、`root_async_client` | LangChain 创建的 openai SDK 客户端对象 |
| `http_client`、`http_async_client`、`http_socket_options` | 自定义底层 HTTP 客户端，如设置代理、证书 |
| `stream_chunk_timeout` | 异步流式输出时，两个数据块之间的最长等待时间 |
| `tiktoken_model_name` | 用 tiktoken 在本地统计 token 数时使用的模型名 |

**OpenAI 专用参数**（了解即可）

`use_responses_api`、`use_previous_response_id`、`reasoning`、`verbosity`、`context_management`、`include`、`prompt_cache_options`、`service_tier`、`store`、`truncation`。这些主要用于 OpenAI 新一代的 Responses API 等 OpenAI 自家功能。因为 ChatDeepSeek 继承自 BaseChatOpenAI，它们才出现在列表中，调用 DeepSeek 时不需要设置。

extra_body: 厂商独有参数。存放模型厂商基于OpenAI API协议扩展的字段。


model_kwargs: 非厂商独有，但在langChain无定义.比如openai兼容协议提供的参数


补充示例（详见 `02-model-init-parameters.ipynb` 第 7 节）：

| 参数 | 放在请求中的位置 | 示例 |
|---|---|---|
| `extra_body` | 原样放进请求体，openai SDK 不检查参数名 | `extra_body={"thinking": {"type": "disabled"}}`（关闭 DeepSeek 思考模式） |
| `model_kwargs` | 合并到请求的顶层参数，openai SDK 会检查参数名 | `model_kwargs={"response_format": {"type": "json_object"}}`（JSON 输出模式） |

如果把厂商独有的参数放进 `model_kwargs`，openai SDK 不认识它，会报错：`TypeError: Completions.create() got an unexpected keyword argument 'thinking'`。

# 9. 调用配置 config 与可配置模型

`04-model-invoke.ipynb` 的 4.1.2 节已经介绍了 config 的基本用法（各个键的含义、run_name / tags / metadata、用 configurable 切换模型），`05-model-stream-batch.ipynb` 介绍了批量调用时的 `max_concurrency`。这里补充几个进阶知识点。

## 9.1 初始化参数与 config 中的同名键

第 8 节的通用参数中有 `tags`、`metadata`、`callbacks`，config 中也有同名的键。区别在于作用范围：

| 位置 | 作用范围 |
|---|---|
| 初始化参数：`init_chat_model(..., tags=[...])` | 这个模型对象的**每一次**调用 |
| config：`model.invoke(..., config={"tags": [...]})` | **本次**调用（以及它内部的子调用） |

两处同时设置时会**合并**：tags 合在一起，metadata 合并成一个字典。如果 metadata 的键重名，以**初始化参数**中的值为准：

```python
from langchain_core.tracers.context import collect_runs
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
load_dotenv(override=True, verbose=True,dotenv_path="/Users/skk/Developer/lang-chain-learn/conf/.env")

model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
    tags=["实例标签"],                          # 初始化参数：每次调用都会带上
    metadata={"app": "学习笔记", "env": "dev"},
)

with collect_runs() as cb:
    model.invoke(
        "你好",
        config={"tags": ["调用标签"], "metadata": {"user_id": "u_001", "env": "prod"}},  # 只对本次调用生效
    )

run = cb.traced_runs[0]
user_metadata = {k: v for k, v in run.metadata.items() if not k.startswith(("ls_", "lc_"))}  # 去掉 LangChain 自动补充的键
print("tags：", run.tags)
print("metadata：", user_metadata)
```

可以看到 tags 合并了，metadata 中重名的 `env` 保留的是初始化参数中的 `dev`，而不是本次调用传入的 `prod`。

## 9.2 with_config()：给模型绑定默认配置

如果每次调用都要传相同的 config，可以用 `with_config()` 预先绑定。它返回一个**新对象**（原模型不受影响），之后每次调用都会自动带上这些配置。

它和 `02-model-init-parameters.ipynb` 第 8 节的 `bind()` 类似：`bind()` 绑定的是模型参数（如 stop），`with_config()` 绑定的是 config。

```python
translator = model.with_config(run_name="翻译助手", tags=["翻译"])  # 返回新对象，原 model 不变
print("类型：", type(translator).__name__)

with collect_runs() as cb:
    translator.invoke("把“早上好”翻译成英文")

run = cb.traced_runs[0]
print("run_name：", run.name)
print("tags：", run.tags)
```

tags 中除了 with_config 绑定的“翻译”，还有“实例标签”：它来自 9.1 中创建 model 时的初始化参数，同样会被合并进来。

## 9.3 可配置模型进阶

04 笔记演示了用 `configurable_fields` 在调用时切换模型，这里补充三点。

**1. 不指定模型，在调用时再决定**

`init_chat_model()` 不传 `model` 时，会得到一个"可配置模型"，`model` 和 `model_provider` 默认就可以在调用时配置。适合模型名来自配置文件或由用户选择的场景。但**调用时必须指定模型**，否则会报错：

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
load_dotenv(override=True, verbose=True,dotenv_path="/Users/skk/Developer/lang-chain-learn/conf/.env")

model = init_chat_model(  # 不传 model
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
)
print("类型：", type(model).__name__)

try:
    model.invoke("你好")
except TypeError as e:
    print("没有指定模型：", e)  # 报错信息不太直观：意思是内部创建模型时缺少 model 参数

response = model.invoke("你好", config={"configurable": {"model": "deepseek:deepseek-v4-flash"}})
print("指定模型后：", response.response_metadata["model_name"], response.content)
```

**2. config_prefix：给可配置字段加前缀**

一个程序里可能有多个可配置模型（比如一个负责写作、一个负责审核），它们的配置键都叫 `model` 就会互相干扰。用 `config_prefix` 给键加上前缀后，配置时要写成 `前缀_字段名`：

```python
writer = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
    configurable_fields=("model",),
    config_prefix="writer",  # 配置键变为 writer_model
)

# 初始化时已经指定了 model_provider，所以这里的模型名不要再加 "deepseek:" 前缀
r1 = writer.invoke("你好")
r2 = writer.invoke("你好", config={"configurable": {"writer_model": "deepseek-v4-pro"}})
r3 = writer.invoke("你好", config={"configurable": {"model": "deepseek-v4-pro"}})  # 没加前缀，不生效
print("默认：", r1.response_metadata["model_name"])
print("writer_model：", r2.response_metadata["model_name"])
print("不加前缀的 model：", r3.response_metadata["model_name"])
```

**3. 慎用 `configurable_fields="any"`**

`configurable_fields="any"` 表示**所有参数**都能在调用时修改，包括 `api_key`、`base_url`。如果 config 的内容来自用户输入（比如 Web 接口的请求参数），别人就可以把请求转发到他们自己的服务器，你的 API Key 也会随请求一起发过去。LangChain 源码注释中特别提醒：接收不可信的配置时，要用 `configurable_fields=(...)` 明确列出允许修改的字段。

## 9.4 小结：模型参数的 5 个设置位置

| 位置 | 写法 | 作用范围 |
|---|---|---|
| 初始化参数 | `init_chat_model(..., temperature=0, tags=[...])` | 该模型对象的每次调用 |
| 调用时传参 | `model.invoke(问题, stop=[...], max_tokens=100)` | 本次调用 |
| config | `model.invoke(问题, config={"tags": [...]})` | 本次调用（及其子调用） |
| bind / with_config | `model.bind(stop=[...])`、`model.with_config(tags=[...])` | 返回的新对象的每次调用 |
| 可配置字段 | `config={"configurable": {"model": ...}}` | 本次调用，需要先用 `configurable_fields` 声明 |

记忆方法：**影响"怎么回答"的模型参数**，用初始化参数、调用时传参或 bind；**追踪信息**（名称、标签、元数据、回调）用 config 或 with_config；**运行时换模型**，用可配置字段。
